# DC Vision Zero Dashboard Backend - Debug Notebook

This notebook helps debug and test the data pipeline step by step.

```python
# Import required libraries
import os
import logging
import pandas as pd
import geopandas as gpd
import requests
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

print("✅ Libraries imported successfully")
```

```python
# Test API connectivity
crashpt_url = "https://maps2.dcgis.dc.gov/dcgis/rest/services/DCGIS_DATA/Public_Safety_WebMercator/MapServer/24/query"

# Test with a small query first
params = {
    "where": "1=1",
    "outFields": "*",
    "outSR": "4326",
    "f": "json",
    "resultOffset": 0,
    "resultRecordCount": 10  # Just get 10 records for testing
}

try:
    response = requests.get(crashpt_url, params=params)
    response.raise_for_status()
    data = response.json()
    print(f"✅ API connection successful")
    print(f"📊 Sample data keys: {list(data.keys())}")
    if 'features' in data:
        print(f"📈 Number of sample features: {len(data['features'])}")
        if data['features']:
            print(f"🔍 Sample feature keys: {list(data['features'][0].keys())}")
except Exception as e:
    print(f"❌ API connection failed: {e}")
```

```python
# Test ArcGIS credentials
client_id = os.environ.get('ARCGIS_CLIENT_ID')
client_secret = os.environ.get('ARCGIS_CLIENT_SECRET')
feature_layer_id = os.environ.get('ARCGIS_FEATURE_LAYER_ID')

if all([client_id, client_secret, feature_layer_id]):
    print("✅ ArcGIS credentials found")
    print(f"🔑 Client ID: {client_id[:10]}...")
    print(f"🗂️ Feature Layer ID: {feature_layer_id}")
    
    try:
        from arcgis.gis import GIS
        gis = GIS("https://dcgis.maps.arcgis.com", client_id=client_id, client_secret=client_secret)
        print("✅ ArcGIS connection successful")
        
        # Test feature layer access
        feature_layer_item = gis.content.get(feature_layer_id)
        if feature_layer_item:
            print(f"✅ Feature layer accessible: {feature_layer_item.title}")
        else:
            print("❌ Feature layer not found")
            
    except Exception as e:
        print(f"❌ ArcGIS connection failed: {e}")
else:
    print("❌ Missing ArcGIS credentials in .env file")
    print("Please check your .env file and ensure all credentials are set")
```

```python
# Check spatial files
spatial_files = [
    'Spatial-Files/crash-hexgrid.geojson',
    'Spatial-Files/anc_2023.geojson',
    'Spatial-Files/smd_2023.geojson',
    'Spatial-Files/Wards_from_2022.geojson'
]

print("📁 Checking spatial files...")
for file_path in spatial_files:
    if os.path.exists(file_path):
        try:
            gdf = gpd.read_file(file_path)
            print(f"✅ {file_path}: {len(gdf)} features")
        except Exception as e:
            print(f"⚠️ {file_path}: Error reading file - {e}")
    else:
        print(f"❌ {file_path}: File not found")

print("\n💡 If spatial files are missing, you'll need to add them to the Spatial-Files/ directory")
```
